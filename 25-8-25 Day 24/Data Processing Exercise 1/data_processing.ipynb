```python
import pandas as pd

# Define sample dataset
data = {
    "TransactionID": [1, 2, 3, 4, 5, 6],
    "CustomerName": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank"],
    "Product": ["Notebook", "Pen", None, "Pencil", "Eraser", "Marker"],
    "Quantity": [2, 5, 3, 10, 1, 2],
    "Price": [5.50, 1.20, 2.00, None, 0.50, 1.75],
    "Date": ["2025-01-05", "2025-02-10", "2025-03-15", "2025-03-20", "2025-04-15", "2025/05/10"]
}

# Create DataFrame
df = pd.DataFrame(data)

# Save as CSV
df.to_csv("raw_sales_data.csv", index=False)

print("raw_sales_data.csv created successfully.")
```

```python
import pandas as pd

# Read raw CSV file
df = pd.read_csv("raw_sales_data.csv")

print("Raw data:")
print(df)
```

```python
import os

# Ensure data folder exists
os.makedirs("data", exist_ok=True)

# Save raw file into data folder
df.to_csv("data/raw_sales_data.csv", index=False)

print("Raw file saved in data/raw_sales_data.csv")
```

```python
# Drop rows with missing values
df_clean = df.dropna()

print("Data after removing missing values:")
print(df_clean)
```

```python
#Convert date columns to YYYY-MM-DD format
from datetime import datetime

def parse_date(val):
    if pd.isna(val):
        return None
    for fmt in ("%Y-%m-%d", "%d-%m-%Y", "%Y/%m/%d"):
        try:
            return datetime.strptime(str(val), fmt).strftime("%Y-%m-%d")
        except ValueError:
            continue
    return None

for col in df_clean.select_dtypes(include=["object", "datetime"]):
    if "date" in col.lower():   # only apply to date-like columns
        df_clean.loc[:, col] = df_clean[col].apply(parse_date)

print("Data after formatting dates:")
print(df_clean)
```

```python
# Convert all column names to lowercase
df_clean.columns = [c.lower() for c in df_clean.columns]

print("Data with lowercase column names:")
print(df_clean.head())
```

```python
# Save cleaned dataset in data folder
df_clean.to_csv("data/clean_sales_data.csv", index=False)

print("Cleaned file saved in data/clean_sales_data.csv")
```
